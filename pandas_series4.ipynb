#### 29. Write a short Pandas program that creates a labelled Series for four products and sales, displays its values and index, selects one label, filters sales above 500, checks whether a label exists, and converts the Series to a dictionary.

```python
sales = Series([450, 650, 800, 300], index=['Pen', 'Notebook', 'Bag', 'Pencil'])

print("Values:")
print(sales.values)

print("Index:")
print(sales.index)

print("Notebook sales:")
print(sales['Notebook'])

print("Sales above 500:")
print(sales[sales > 500])

print("Is Bag present?")
print('Bag' in sales.index)

print("Dictionary:")
print(sales.to_dict())
```

#### 30. Analyse the following code and answer all parts: (a) What is the index order of s? (b) Which label contains NaN? (c) What will the Boolean filter s[s &gt; 20] return? (d) What will pd.notnull(s) indicate for each label? (e) What are the Series name and index name after the final two statements?

```python
data = {'A': 15, 'B': 35, 'C': 25}

s = Series(data, index=['C', 'A', 'D', 'B'])

s.name = 'Observations'
s.index.name = 'Category'

print(s)
print(s[s > 20])
print(pd.notnull(s))
print(s.name)
print(s.index.name)
```
